# 🧠 Single Agent Pipeline Project

## Problem Statement
Build a **Single-Agent Smart Assistant** that:
- Understands user queries
- Routes tasks based on intent
- Uses tools when required
- Returns structured JSON output

### The agent should handle:
- Math queries → Calculator Tool
- Keyword extraction → Keyword Tool
- General queries → Direct response

---
### 🛠️ What You Need to Implement
- Agent logic
- Conditional routing
- Tool integration
- Basic error handling

### 🚀 Bonus
- Improve routing
- Add logging
- Add more tools


````python
# 🛠️ TOOL 1: Calculator

def calculator(expression: str) -> str:
    """Evaluate a mathematical expression."""
    try:
        return str(eval(expression))
    except Exception:
        return "Error in calculation"
````

````python
# 🛠️ TOOL 2: Keyword Extractor

def extract_keywords(text: str) -> list:
    """Extract keywords from text."""
    try:
        words = text.split()
        keywords = list(set([w.lower() for w in words if len(w) > 4]))
        return keywords[:5]
    except Exception:
        return []
````

# Bonus Tool

````python
# BONUS TOOL: Word Counter

def word_counter(text: str):
    """Count total words in a sentence."""

    return len(text.split())
````

## 🤖 Implement Agent Logic Below

👉 Use conditional routing:
- If query contains "calculate" → use calculator
- If query contains "keywords" → use keyword extractor
- Else → general response

## Enhanced Agent Implementation

The agent has been enhanced with:

- Improved intent routing
- Logging for request processing
- Additional Word Counter Tool
- JSON formatted responses
- Error handling

````python
# AGENT FUNCTION

def agent(query: str):

    query_lower = query.lower()

    print(f"[LOG] Query Received: {query}")

    try:

        # Calculator Tool
        if any(word in query_lower for word in ["calculate", "compute", "solve", "math"]):

            print("[LOG] Intent Detected: Calculation")
            print("[LOG] Using Calculator Tool")

            expression = query_lower

            for word in ["calculate", "compute", "solve", "math"]:
                expression = expression.replace(word, "")

            expression = expression.strip()

            result = calculator(expression)

            if result == "Error in calculation":
                return {
                    "type": "error",
                    "result": "Invalid mathematical expression."
                }

            return {
                "type": "calculation",
                "result": result
            }

        # Keyword Extraction Tool
        elif "keyword" in query_lower:

            print("[LOG] Intent Detected: Keyword Extraction")
            print("[LOG] Using Keyword Extractor")

            text = query

            phrases = [
                "extract keywords from",
                "find keywords from",
                "keywords"
            ]

            for phrase in phrases:
                text = text.replace(phrase, "")
                text = text.replace(phrase.title(), "")

            result = extract_keywords(text.strip())

            return {
                "type": "keywords",
                "result": result
            }
        # Word Counter Tool (Bonus)
        elif "count words" in query_lower:

            print("[LOG] Intent Detected: Word Count")
            print("[LOG] Using Word Counter Tool")

            text = query_lower.replace("count words", "").strip()

            return {
                "type": "word_count",
                "result": word_counter(text)
            }

        # General Response
        else:

            print("[LOG] Intent Detected: General Query")

            return {
                "type": "general",
                "result": "Hello! I can perform calculations, extract keywords, and count words."
            }

    except Exception as e:

        print("[LOG] Error:", e)

        return {
            "type": "error",
            "result": str(e)
        }
````

## 📦 Expected Output Format

```
{
  "type": "calculation / keywords / general / error",
  "result": ...
}
```

````python
# 🧪 Test Cases

queries = [
    "Calculate 20 + 5",
    "Extract keywords from Artificial Intelligence is transforming industries",
    "Count words Artificial Intelligence is transforming healthcare",
    "What is machine learning?",
    "Calculate 20 +"
]

for q in queries:
    print("Query:", q)
    print("Response:", agent(q))
    print("-" * 50)
````

````python
# 🎯 Interactive Mode

while True:
    user_input = input("Enter query (type 'exit' to stop): ")
    if user_input.lower() == "exit":
        break
    print("Response:", agent(user_input))
````
